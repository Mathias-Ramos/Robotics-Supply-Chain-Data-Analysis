```python
import pandas as pd
```

```python
# Load the CSV file in a DataFrame.
df = pd.read_csv('../data/processed/stocks_cleaned.csv')
df
```

# Transformation

Only keep the layer id in the Layer column.

```python
df["layer_id"] = df["layer"].str.findall(r"(\d+)")
df = df.drop(["layer"], axis=1)
df.head(30)
```

Group the duplicate tickers.

```python
df_grouped = df.groupby(["ticker"], as_index=False)[["layer_id"]].agg({"layer_id": "first"})
df_grouped
```

Check for any duplicates left.

```python
df_grouped[df_grouped["ticker"].duplicated(keep=False)]
```

# Export

```python
df.to_csv('../data/processed/stocks_transformed.csv', index=False)
```
