```python
import pandas as pd
import plotly.express as px
```

```python
# Load KPI data
df_kpis = pd.read_csv("../data/processed/stocks_kpis.csv")
```

Create the graph:

```python
# Keep companies with a meaningful PEG ratio
df_chart = df_kpis[df_kpis["PEG Ratio (1Y)"] > 0]

# Remove extreme PEG outliers
df_chart = df_chart[df_chart["PEG Ratio (1Y)"] <= 10]

# Create scatter plot
fig = px.scatter(
    df_chart,
    x="PEG Ratio (1Y)",
    y="ROIC (%)",
    text="Ticker",
    hover_data=["Name"]
)

fig.show()
```

# How to read the graph?

- **Upper-left:** High ROIC + Low PEG = efficient companies at relatively attractive valuations.
- **Upper-right:** High ROIC + High PEG = strong companies, but relatively expensive.
- **Lower-left:** Low ROIC + Low PEG = relatively cheap, but lower capital efficiency.
- **Lower-right:** Low ROIC + High PEG = lower efficiency and relatively expensive.
