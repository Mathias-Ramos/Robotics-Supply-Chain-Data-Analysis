```python
import pandas as pd
```

```python
# Load the CSV file in a DataFrame.
df = pd.read_csv('../data/raw/supply_chain_levels.csv')
df
```

# Cleaning

Separate *Supply Chain Level* colomn in 2: the supply chain level (number) and its name.

```python
df['chain_level_nb'] = df['Supply Chain Level'].str.extract(r'(\d+)').astype(int)
df['chain_level'] = df['Supply Chain Level'].str.extract(r'\d+\.\s*(.*)')
df
```

Rename the description column to lower case.

```python
df = df.rename(columns={'Description' : 'description'})
df
```

Export a clean file.

```python
df_clean = df[['chain_level_nb', 'description', 'chain_level']].copy()
df_clean
```

```python
df_clean.to_csv('../data/processed/supply_chain_levels_clean.csv', index=False)
```
